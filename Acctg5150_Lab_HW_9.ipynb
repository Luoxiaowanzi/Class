```python

# Load Imports
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf

df = pd.read_csv('production_costs_daily.csv', parse_dates=['date'])

# View first 2 rows
df.head(2)
```

```python

# Run simple regression: total_cost ~ volume
exercise1_volume_model = smf.ols('total_cost ~ volume', data=df).fit()

# Output regression output for submission screenshot
print(exercise1_volume_model.summary())
```

```python

from sklearn.metrics import mean_squared_error

df['total_cost_hat_e1'] = exercise1_volume_model.predict(df)

rmse_e1 = np.sqrt(mean_squared_error(df['total_cost'], df['total_cost_hat_e1']))

print(f"RMSE: ${rmse_e1:,.2f}")
```

```python

from sklearn.metrics import mean_absolute_error

mae_e1 = mean_absolute_error(df['total_cost'], df['total_cost_hat_e1'])

print(f"MAE: ${mae_e1:,.2f}")
```

```python

from sklearn.metrics import mean_absolute_percentage_error

# Calculate MAPE for Exercise 1
df['total_cost_hat_e1'] = exercise1_volume_model.predict(df)

mape_e1 = mean_absolute_percentage_error(df['total_cost'], df['total_cost_hat_e1']) * 100

print(f"MAPE: {mape_e1:.2f}%")
```

```python

# Run ABC regression model
exercise2_abc_model = smf.ols('total_cost ~ material_lbs + labor_hours + machine_hours + num_batches + num_inspections', data=df).fit()

# Output regression results
print(exercise2_abc_model.summary())
```

```python

# Calculate predictions for Exercise 2
df['total_cost_hat_e2'] = exercise2_abc_model.predict(df)

# Calculate RMSE
rmse_e2 = np.sqrt(mean_squared_error(df['total_cost'], df['total_cost_hat_e2']))

print(f"RMSE: ${rmse_e2:,.2f}")
```

```python

# Calculate MAE for Exercise 2
mae_e2 = mean_absolute_error(df['total_cost'], df['total_cost_hat_e2'])

print(f"MAE: ${mae_e2:,.2f}")
```

```python

from sklearn.metrics import mean_absolute_percentage_error

# Calculate MAPE for Exercise 2
df['total_cost_hat_e2'] = exercise2_abc_model.predict(df)

mape_e2 = mean_absolute_percentage_error(df['total_cost'], df['total_cost_hat_e2']) * 100

print(f"MAPE: {mape_e2:.2f}%")
```

```python

# Exercise 3: Volume with Quarter Fixed Effects
exercise3_qfe_model = smf.ols('total_cost ~ volume + C(quarter)', data=df).fit()

# Output regression results
print(exercise3_qfe_model.summary())
```

```python

from sklearn.metrics import mean_squared_error
import numpy as np

# Calculate predictions for Exercise 3
df['total_cost_hat_e3'] = exercise3_qfe_model.predict(df)

# Calculate RMSE
rmse_e3 = np.sqrt(mean_squared_error(df['total_cost'], df['total_cost_hat_e3']))

print(f"RMSE: ${rmse_e3:,.2f}")
```

```python

from sklearn.metrics import mean_absolute_error

# Calculate predictions for Exercise 3
df['total_cost_hat_e3'] = exercise3_qfe_model.predict(df)

# Calculate MAE
mae_e3 = mean_absolute_error(df['total_cost'], df['total_cost_hat_e3'])

print(f"MAE: ${mae_e3:,.2f}")
```

```python

from sklearn.metrics import mean_absolute_percentage_error

# Calculate predictions for Exercise 3
df['total_cost_hat_e3'] = exercise3_qfe_model.predict(df)

# Calculate MAPE
mape_e3 = mean_absolute_percentage_error(df['total_cost'], df['total_cost_hat_e3']) * 100

print(f"MAPE: {mape_e3:.2f}%")
```

```python

# Exercise 4: Train/Test Split

# Train volume model on 2015-2022 data only
model_train = smf.ols('total_cost ~ volume', data=df[df['year'] < 2023]).fit()

# Output regression results
print(model_train.summary())
```

```python

from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_absolute_percentage_error
import numpy as np
import pandas as pd

# Predict using the training model
df['total_cost_hat'] = model_train.predict(df)

# Calculate error metrics
def calc_metrics(data):
    actual = data['total_cost']
    predicted = data['total_cost_hat']

    rmse = np.sqrt(mean_squared_error(actual, predicted))
    mae = mean_absolute_error(actual, predicted)
    mape = mean_absolute_percentage_error(actual, predicted) * 100

    return {
        'RMSE': rmse,
        'MAE': mae,
        'MAPE (%)': mape,
        'N': len(data)
    }

# Training and testing results
train_metrics = calc_metrics(df[df['year'] < 2023])
test_metrics = calc_metrics(df[df['year'] >= 2023])

pd.DataFrame(
    [train_metrics, test_metrics],
    index=['Training (2015-2022)', 'Testing (2023-2024)']
).round(2)
```
