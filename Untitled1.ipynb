```python
import pandas as pd
df = pd.read_excel("lab7_data.xlsx")
df.head()
```

```python

df['Invoice_Number'] = df['notes'].str.extract(
    r"Inv\w*\s*#:\s*([^\n]+)",
    expand=False
)

df[['notes', 'Invoice_Number']].head()
```

```python

df['Contact_Name'] = df['notes'].str.extract(
    r"(?:A\w*\s*)?Rep\w*:\s*([^\n]+)",
    expand=False
)

df[['notes', 'Contact_Name']].head()
```

```python

df['Follow_Up_Date'] = df['notes'].str.extract(
    r"Follow[\w\s-]*:\s*([^\n]+)",
    expand=False
)

df[['notes', 'Follow_Up_Date']].head()
```

```python

df['Clean_Notes'] = df['notes'].str.extract(
    r"Notes?:\s*([\s\S]+)",
    expand=False
)

df[['notes', 'Clean_Notes']].head()
```

```python

df.to_csv("lab7_data_cleaned.csv", index=False)
```
